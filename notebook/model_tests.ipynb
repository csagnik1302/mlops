```python
from xgboost import XGBClassifier
import lightgbm as lgb
from catboost import CatBoostClassifier
from pathlib import Path
import pandas as pd
from optuna.integration import OptunaSearchCV
from optuna.distributions import IntDistribution, FloatDistribution
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import make_scorer, recall_score, classification_report, f1_score
from sklearn.utils.class_weight import compute_sample_weight
import json

notebook_path=Path('__file__').resolve().parent
ROOT=notebook_path.parent
```

```python
PATH_TRAIN_PARSED=ROOT/'data'/'parsed'/'train.csv'
PATH_TEST_PARSED=ROOT/'data'/'parsed'/'test.csv'
PATH_TRAIN_LABEL_PARSED=ROOT/'data'/'parsed'/'train_label.csv'
PATH_TEST_LABEL_PARSED=ROOT/'data'/'parsed'/'test_label.csv'

PATH_OPTUNA=ROOT/'data'/'model'/'optuna_data'
PATH_MODEL_PF=ROOT/'data'/'model'/'model_performance'
PATH_MODEL_PARS=ROOT/'data'/'model'/'model_optimal_parameters'

train_data=pd.read_csv(PATH_TRAIN_PARSED)
test_data=pd.read_csv(PATH_TEST_PARSED)
train_label=pd.read_csv(PATH_TRAIN_LABEL_PARSED).squeeze()
test_label=pd.read_csv(PATH_TEST_LABEL_PARSED).squeeze()
```

```python
import numpy as np

sample_weight=compute_sample_weight(class_weight='balanced', y=train_label)    # weight is assigned inversely proportional to class frequency
np.unique(sample_weight)
```

```python
def run_optuna(train_data, train_label, test_data, test_label, base_model, n_trials, param_distributions, model_name_for_export):

    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    recall_scorer=make_scorer(f1_score, average='macro')

    optuna_search=OptunaSearchCV(estimator=base_model,
                                param_distributions=param_distributions,
                                cv=cv,
                                scoring=recall_scorer,
                                n_trials=n_trials,
                                random_state=42)

    sample_weight=compute_sample_weight(class_weight='balanced', y=train_label)

    optuna_search.fit(train_data, train_label, sample_weight=sample_weight)

    with open(PATH_MODEL_PARS/f"{model_name_for_export}.json", 'w') as f:
        json.dump(optuna_search.best_params_,f)

    test_pred=optuna_search.predict(test_data)
    report=classification_report(y_true=test_label, y_pred=test_pred, output_dict=True)
    report_df=pd.DataFrame(report).T
    report_df.to_csv(PATH_MODEL_PF/f"{model_name_for_export}_model_performance.csv", index=True)

    df = optuna_search.study_.trials_dataframe()
    df.to_csv(PATH_OPTUNA/f"{model_name_for_export}_study_results.csv", index=False)

    return report_df
```

### Model Data

```python
xgb_model=XGBClassifier(random_state=42)

xgb_distributions={'n_estimators': IntDistribution(50, 300),
                    'learning_rate': FloatDistribution(0.01, 0.5, log=True),
                    'max_depth': IntDistribution(3, 10),
                    'subsample': FloatDistribution(0.5, 1.0),
                    'colsample_bytree': FloatDistribution(0.5, 1.0),
                    'gamma': FloatDistribution(0, 5),
                    'reg_alpha': FloatDistribution(1e-3, 1.0, log=True),
                    'reg_lambda': FloatDistribution(1.0, 11.0),}



lgbm_model=lgb.LGBMClassifier(random_state=42)

lgbm_distributions={'n_estimators': IntDistribution(50, 300),
                    'learning_rate':     FloatDistribution(0.01, 0.5, log=True),
                    'num_leaves':        IntDistribution(15, 255),          # LightGBM's main complexity control (leaf-wise growth)
                    'max_depth':         IntDistribution(3, 12),            # -1 (unlimited) is also valid but risky to search over directly
                    'min_child_samples': IntDistribution(5, 100),           # LightGBM's equivalent of min_samples_leaf; important for small/imbalanced classes
                    'subsample':         FloatDistribution(0.5, 1.0),
                    'colsample_bytree':  FloatDistribution(0.5, 1.0),
                    'reg_alpha':         FloatDistribution(1e-3, 1.0, log=True),
                    'reg_lambda':        FloatDistribution(1e-3, 10.0, log=True),}


cat_model=CatBoostClassifier(random_state=42)

cat_distributions={'iterations':         IntDistribution(50, 300),                              # CatBoost's name for n_estimators
                    'learning_rate':      FloatDistribution(0.01, 0.5, log=True),
                    'depth':              IntDistribution(3, 10),                              # CatBoost's max_depth equivalent, symmetric trees
                    'l2_leaf_reg':        FloatDistribution(1.0, 10.0, log=True),              # CatBoost's main regularization param
                    'random_strength':    FloatDistribution(1e-3, 10.0, log=True),              # randomness in scoring splits, helps avoid overfitting
                    'bagging_temperature': FloatDistribution(0.0, 1.0)}                         # controls Bayesian bootstrap intensity
```

```python
xgb_out=run_optuna(train_data=train_data, train_label=train_label, test_data=test_data, test_label=test_label, base_model=xgb_model, n_trials=30, param_distributions=xgb_distributions, model_name_for_export='xgboost')
lgb_out=run_optuna(train_data=train_data, train_label=train_label, test_data=test_data, test_label=test_label, base_model=lgbm_model, n_trials=30, param_distributions=lgbm_distributions, model_name_for_export='lightgbm')
cat_out=run_optuna(train_data=train_data, train_label=train_label, test_data=test_data, test_label=test_label, base_model=cat_model, n_trials=30, param_distributions=cat_distributions, model_name_for_export='catboost')
```
